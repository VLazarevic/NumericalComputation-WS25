## Exercise 4.1 – Newton–Cotes Properties

Let the (closed) Newton–Cotes formula on [0,1] be

$$
\widehat Q^{NC}_n(f) = \sum_{i=0}^n w_i f(x_i),
\quad \text{where } x_i = \frac{i}{n}.
$$

(The open Newton–Cotes rule $\widehat Q^{oNC}_n$ uses $x_i = \frac{i}{n+1},\ i=1,\dots,n.$)

---

### a) Show that $\sum_{i=0}^n w_i = 1$

Apply the quadrature formula to the constant function $f(x) = 1$:

$$
\int_0^1 1\,dx = 1 = \widehat Q^{NC}_n(1) = \sum_{i=0}^n w_i f(x_i) = \sum_{i=0}^n w_i.
$$

Hence:

$$
\sum_{i=0}^n w_i = 1.
$$

(Similarly, for the open rule $\sum_{i=1}^n w_i = 1$.)

---

### b) Show the symmetry property $w_{n-i} = w_i$

Each weight can be expressed as an integral of the Lagrange basis polynomial:

$$
w_i = \int_0^1 \ell_i(x)\,dx,
\quad \text{where } \ell_i(x_j) = \delta_{ij}.
$$

Because the nodes are symmetric about $1/2$, we have $x_{n-j} = 1 - x_j$.
This implies $\ell_{n-i}(x) = \ell_i(1 - x)$.

Now perform the change of variable $u = 1 - x$:

$$
w_{n-i} = \int_0^1 \ell_{n-i}(x)\,dx
= \int_0^1 \ell_i(1 - x)\,dx
= \int_0^1 \ell_i(u)\,du
= w_i.
$$

Therefore $w_{n-i} = w_i$.

---

### c) Case $n = 2m$ (even)

Let

$$
f(x) = \left(x - \frac{1}{2}\right)^{n+1}.
$$

Then $f$ is **antisymmetric** with respect to $x = \frac{1}{2}$:
$$
f(1 - x) = -f(x).
$$

Hence:
$$
\int_0^1 f(x)\,dx = 0.
$$

Using the symmetry of weights and nodes:

$$
\begin{aligned}
\widehat Q^{NC}_n(f)
&= \sum_{i=0}^n w_i f(x_i)
= w_m f\!\left(\frac{1}{2}\right)
+ \sum_{i=0}^{m-1} [w_i f(x_i) + w_{n-i} f(x_{n-i})] \\
&= 0 + \sum_{i=0}^{m-1} w_i [f(x_i) + f(1 - x_i)] = 0.
\end{aligned}
$$

The same holds for the open Newton–Cotes formula:
$$
\widehat Q^{oNC}_n(f) = 0.
$$

Since both formulas are exact for all polynomials of degree ≤ n (by construction)
and also for one polynomial of degree n+1, we conclude:

$$
\text{For even } n,\;
\widehat Q^{NC}_n \text{ and } \widehat Q^{oNC}_n
\text{ are exact for all polynomials of degree } \le n + 1.
$$

## 4.2 — Extrapolation of the composite trapezoidal rule equals the composite Simpson rule

Let:
- \(T(h)\) be the composite trapezoidal rule on \([a,b]\) with step \(h=(b-a)/N\),
- \(S(h)\) be the composite Simpson rule with the same step \(h\),
- Romberg step sizes \(h_i=(b-a)2^{-i}\) so that \(h_{i+1}=h_i/2\).

---

### a) First Richardson/Neville column \(m=1\) for \(T(h)\)

For smooth \(f\), the error expansion of the composite trapezoidal rule is
$$
T(h)=I+c_2h^2+c_4h^4+\cdots,\qquad
T(h/2)=I+c_2\frac{h^2}{4}+c_4\frac{h^4}{16}+\cdots .
$$

Eliminate the \(O(h^2)\) term:
$$
N(h)\equiv\frac{4\,T(h/2)-T(h)}{3}
= T(h/2)+\frac{T(h/2)-T(h)}{3}=I+O(h^4).
$$

With \(h=h_i\) and \(h/2=h_{i+1}\) this gives the \(m=1\) Neville (Romberg) entries
$$
N_i := T(h_{i+1})+\tfrac{1}{3}\big(T(h_{i+1})-T(h_i)\big).
$$

---

### b) Show that $(N_i=S(h_i))$

Write the trapezoidal rules on coarse and fine grids:
$$
\begin{aligned}
T(h_i)
&= h_i\Big(\tfrac12 f(a)+\sum_{k=1}^{2^{\,i}-1} f(a+k h_i)+\tfrac12 f(b)\Big),\\[2mm]
T(h_{i+1})
&= \frac{h_i}{2}\Big(\tfrac12 f(a)+\sum_{k=1}^{2^{\,i+1}-1} f\!\big(a+k\tfrac{h_i}{2}\big)+\tfrac12 f(b)\Big).
\end{aligned}
$$

Combine as in part (a):
$$
\frac{4T(h_{i+1})-T(h_i)}{3}
= \frac{h_i}{3}\!\left[
f(a)
+4\sum_{\ell=0}^{2^{\,i}-1} f\!\Big(a+(2\ell+1)\tfrac{h_i}{2}\Big)
+2\sum_{\ell=1}^{2^{\,i}-1} f(a+\ell h_i)
+f(b)
\right].
$$

The right-hand side is exactly the composite Simpson rule with step \(h_i\):
$$
S(h_i)=\frac{h_i}{3}\left[
f(a)
+4\sum_{\ell=0}^{2^{\,i}-1} f\!\Big(a+(2\ell+1)\tfrac{h_i}{2}\Big)
+2\sum_{\ell=1}^{2^{\,i}-1} f(a+\ell h_i)
+f(b)
\right].
$$

Therefore,
$$
N_i=\frac{4T(h_{i+1})-T(h_i)}{3}=S(h_i).
$$

**Conclusion.** The first Romberg extrapolation of the composite trapezoidal rule equals the composite Simpson rule.

## 4.3 — Quadrature rule
Given
$$
\int_{0}^{1} f(x)\,dx \;\approx\; \tfrac12\,f\!\left(\tfrac13\right)+\tfrac12\,f\!\left(\tfrac23\right).
$$

---

### a) Evaluate for $(f(x)=3x^2+1)$ and compute the error

Exact integral:
$$
\int_0^1 (3x^2+1)\,dx
= 3\!\int_0^1 x^2\,dx + \int_0^1 1\,dx
= 3\cdot\frac13 + 1 = 2.
$$

Quadrature value:
$$
f\!\left(\tfrac13\right)=3\cdot\frac{1}{9}+1=\frac{4}{3}, \qquad
f\!\left(\tfrac23\right)=3\cdot\frac{4}{9}+1=\frac{7}{3}.
$$
Hence
$$
Q=\tfrac12\!\left(\frac{4}{3}\right)+\tfrac12\!\left(\frac{7}{3}\right)
= \frac{11}{6}\approx 1.8333\ldots
$$

Error (exact − quadrature):
$$
E = 2 - \frac{11}{6} = \frac{1}{6}\quad\text{(the rule underestimates by }1/6\text{)}.
$$

---

### b) Maximal degree of exactness

Test monomials:

- $(f(x)=1)$: $$(Q= \tfrac12+\tfrac12=1=\int_0^1 1\,dx)$$
- $(f(x)=x)$: $$(Q=\tfrac12(\tfrac13+\tfrac23)=\tfrac12=\int_0^1 x\,dx)$$
- $(f(x)=x^2)$: $$(Q=\tfrac12\!\left(\tfrac{1}{9}+\tfrac{4}{9}\right)=\tfrac{5}{18}\neq \frac13)$$

So the rule is exact for degrees \(0\) and \(1\) but not for degree \(2\).
$$
\boxed{\text{Degree of exactness }=1}
$$

---

### c) Can different weights or knots give higher exactness?

- **Different weights with the *same* knots $(\tfrac13,\tfrac23)$**:  
  Let weights be $(a,b)$. Exactness for $(1)$ and $(x)$ gives
  $(a+b=1)$ and $(\tfrac13 a+\tfrac23 b=\tfrac12)$, yielding $(a=b=\tfrac12)$.  
  Then for $(x^2)$: $(\tfrac19 a+\tfrac49 b=\tfrac{5}{18}\neq \tfrac13)$.  
  Thus **no choice of weights** (with these fixed knots) raises the degree.

- **Different knots**:  
  With **two nodes**, the best possible degree is \(3\) (Gauss–Legendre, mapped to \([0,1]\)).  
  The Gauss nodes on \([0,1]\) are
  $$
  x_{1,2}=\frac12 \pm \frac{1}{2\sqrt{3}}, \qquad \text{with equal weights } \frac12,
  $$
  which integrate all polynomials up to degree \(3\) exactly.

**Conclusion:** You cannot improve the degree by changing only the weights at $(1/3,\,2/3)$.  
To achieve a higher degree (up to \(3\) with two nodes), you must **move the knots** to the Gauss points $(\tfrac12 \pm \tfrac{1}{2\sqrt{3}})$ (weights $(1/2)$).

```python
# Adaptive Simpson integration for Exercise 4.4

import math
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# --- Simpson on one interval [a,b]
def simpson(f, a, b):
    m = 0.5*(a+b)
    h = b - a
    return (h/6.0) * (f(a) + 4.0*f(m) + f(b))

# --- Adaptive Simpson based on comparing S(a,b) vs S(a,m)+S(m,b)
def adapt(f, a, b, tau, hmin):
    """
    Adaptive Simpson rule.
    - tau   : desired absolute accuracy (local tolerance)
    - hmin  : minimal interval length (stopping condition)
    Returns: approximation of \int_a^b f(x) dx
    """
    S = simpson(f, a, b)
    m = 0.5*(a+b)
    S_left  = simpson(f, a, m)
    S_right = simpson(f, m, b)
    S2 = S_left + S_right
    # Richardson error estimate for Simpson
    err_est = abs(S2 - S) / 15.0

    # accept if accurate enough or cannot split further
    if err_est <= tau or (b - a) <= hmin:
        return S2 + (S2 - S) / 15.0  # corrected value

    # otherwise split and recurse (split tolerance between children)
    return adapt(f, a, m, tau/2.0, hmin) + adapt(f, m, b, tau/2.0, hmin)

# --- Test function from the exercise
def f(x):
    return 0.5*math.exp(x) if x < 1.0/3.0 else math.exp(x)

# exact value: ∫_0^{1/3} 0.5 e^x dx + ∫_{1/3}^1 e^x dx = e - 0.5 e^{1/3} - 0.5
I_exact = math.e - 0.5*math.exp(1.0/3.0) - 0.5

# Run experiment for tau = hmin = 2^{-j},  j=0,...,10
rows = []
for j in range(0, 10):
    tau = 2.0**(-j)
    I = adapt(f, 0.0, 1.0, tau=tau, hmin=tau)  # hmin = tau (as requested)
    err = abs(I - I_exact)
    rows.append((j, tau, I, err))

df = pd.DataFrame(rows, columns=["j", "tau=hmin", "I_adapt", "abs_error"])
print("Exact integral I =", I_exact)
df

# Plot error vs tau (log-log)
plt.figure()
plt.loglog(df["tau=hmin"], df["abs_error"], marker="o")
plt.xlabel("tau ( = h_min )")
plt.ylabel("absolute error")
plt.title("Adaptive Simpson: error vs tau for discontinuous f")
plt.grid(True, which="both")
plt.show()

# Estimate order on the finest points (where refinement has already located the jump)
lo = 7  # use the last few data points
p = np.polyfit(np.log(df["tau=hmin"].values[lo:]), np.log(df["abs_error"].values[lo:]), 1)
print("Observed slope (order wrt tau) on fine grid ≈", p[0])
```

- **Why choose** $h_{\min} = \tau$ ?

  The integrand has a **jump discontinuity** at $x = \tfrac{1}{3}$.  
  The adaptive Simpson method estimates local error by comparing  
  $S_{[a,b]}(f)$ with $S_{[a,m]}(f) + S_{[m,b]}(f)$.  
  Near a discontinuity, this difference never gets small — so the algorithm
  would keep subdividing indefinitely unless we impose a lower limit on interval size.  

  Setting $h_{\min} = \tau$ ensures:
  - The recursion **terminates** safely.
  - The smallest local error and the smallest interval are of the **same order**,
    so refinement stops naturally where it should.

---

- **Observed convergence behavior**

  For **coarse tolerances** (large $\tau$), the algorithm doesn’t yet refine near the
  discontinuity, so the **error plateaus** (stays roughly constant).

  Once $\tau$ becomes small enough to resolve the jump,
  the **log–log plot of error vs. $\tau$** becomes a straight line with slope ≈ 1.

  Hence, the **observed convergence rate is**
  $$
  \text{error} = O(\tau^1).
  $$

  This linear rate occurs because the dominant error comes from intervals
  near the discontinuity, where the integration accuracy scales linearly
  with the local interval size. Since $h_{\min} = \tau$, we get
  $$
  \text{global error} = O(h_{\min}) = O(\tau).
  $$

---

- **Final value (for $j=10$):**

  $$
  I_{\text{adapt}} \approx 1.5204074793613827, \qquad
  \text{absolute error} \approx 6.81\times10^{-5}.
  $$

  The **exact value** of the integral is
  $$
  I_{\text{exact}} = e - \tfrac{1}{2} e^{1/3} - \tfrac{1}{2}
  \;\approx\; 1.5204756159160002.
  $$

---

**Summary:**
- The adaptive Simpson rule correctly refines near the discontinuity.  
- With $h_{\min}=\tau$, the algorithm’s stopping criterion is consistent with its accuracy.  
- The observed convergence is **first order**, i.e. the error decreases linearly with $\tau$.
